```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
import pandas as pd

data = {
    "Student_ID": range(1, 21),

    "Gender": [
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],

    "Age": [
        20, 21, 20, 22, 21,
        20, 22, 21, 20, 23,
        21, 22, 20, 21, 22,
        20, 21, 23, 20, 22
    ],

    "Attendance": [
        85, 78, 92, 74, 88,
        69, 95, 81, 90, 72,
        87, 76, 93, 80, 89,
        70, 84, 91, 77, 86
    ],

    "Study_Hours": [
        4, 3, 5, 2, 4,
        2, 6, 4, 5, 3,
        4, 3, 6, 4, 5,
        2, 4, 5, 3, 4
    ],

    "Math_Marks": [
        85, 72, 91, 65, 88,
        60, 95, 78, 90, 68,
        84, 70, 93, 76, 87,
        62, 82, 89, 71, 85
    ],

    "Science_Marks": [
        82, 75, 89, 68, 90,
        64, 94, 80, 88, 70,
        86, 73, 91, 78, 85,
        66, 84, 92, 74, 87
    ],

    "English_Marks": [
        88, 79, 92, 70, 86,
        67, 96, 82, 91, 73,
        89, 76, 94, 80, 88,
        69, 85, 90, 77, 86
    ]
}

df = pd.DataFrame(data)

df.to_csv("student_performance.csv", index=False)

print("Student dataset created successfully!")
print("Dataset saved as student_performance.csv")
```

```python
print("First 5 rows of the dataset:")
display(df.head())
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("Statistical Summary:")

display(df.describe())
```

```python
subject_averages = df[
    ["Math_Marks", "Science_Marks", "English_Marks"]
].mean()

print("Average Marks:")
print(subject_averages)
```

```python
high_attendance = df[df["Attendance"] > 85]

print("Students with attendance above 85%:")
display(high_attendance)
```

```python
math_sorted = df.sort_values(
    by="Math_Marks",
    ascending=False
)

print("Students sorted by Mathematics marks:")
display(math_sorted)
```

```python
gender_analysis = df.groupby("Gender")[
    [
        "Attendance",
        "Study_Hours",
        "Math_Marks",
        "Science_Marks",
        "English_Marks"
    ]
].mean()

print("Average Performance by Gender:")
display(gender_analysis)
```

```python
df["Average_Marks"] = df[
    ["Math_Marks", "Science_Marks", "English_Marks"]
].mean(axis=1)

print("Average_Marks column created successfully!")

display(
    df[
        [
            "Student_ID",
            "Math_Marks",
            "Science_Marks",
            "English_Marks",
            "Average_Marks"
        ]
    ].head()
)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Average_Marks"],
    bins=6,
    edgecolor="black"
)

plt.title("Distribution of Student Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Gender")

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

```python
subject_averages = df[
    ["Math_Marks", "Science_Marks", "English_Marks"]
].mean()

plt.figure(figsize=(8, 5))

subject_averages.plot(kind="bar")

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance"],
    df["Average_Marks"],
    s=80
)

plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9, 6))

correlation = df[
    [
        "Attendance",
        "Study_Hours",
        "Math_Marks",
        "Science_Marks",
        "English_Marks",
        "Average_Marks"
    ]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap of Student Performance")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Average_Marks"],
    s=80
)

plt.title("Study Hours vs Average Marks")
plt.xlabel("Study Hours per Day")
plt.ylabel("Average Marks")

plt.grid(True)
plt.show()
```

```python
print("===== STUDENT PERFORMANCE ANALYSIS =====\n")

print("1. Subject-wise Average Marks:")
print(subject_averages)

highest_subject = subject_averages.idxmax()
highest_average = subject_averages.max()

print(
    f"\n2. Highest Average Performance: "
    f"{highest_subject} ({highest_average:.2f} marks)"
)

attendance_correlation = df["Attendance"].corr(
    df["Average_Marks"]
)

print(
    f"\n3. Attendance vs Average Marks Correlation: "
    f"{attendance_correlation:.2f}"
)

study_correlation = df["Study_Hours"].corr(
    df["Average_Marks"]
)

print(
    f"4. Study Hours vs Average Marks Correlation: "
    f"{study_correlation:.2f}"
)

top_student = df.loc[
    df["Average_Marks"].idxmax()
]

print(
    f"\n5. Highest Performing Student: "
    f"Student {int(top_student['Student_ID'])} "
    f"with {top_student['Average_Marks']:.2f} average marks"
)
```
